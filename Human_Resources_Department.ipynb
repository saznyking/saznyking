```python
x_cat = data.select_dtypes(include = ['object', 'category'])
```

```python
x_cat.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
onehotencoder = OneHotEncoder()

cat_cols = onehotencoder.fit_transform(x_cat).toarray()
```

```python
cat_cols.shape
```

```python
cat_cols = pd.DataFrame(cat_cols, columns = onehotencoder.get_feature_names_out(x_cat.columns))
```

```python
cat_cols.head()
```

```python
num_cols = data.select_dtypes(include = ['number'])
```

```python
num_cols.head()
```

```python
new_df = pd.concat([num_cols, cat_cols], axis = 1)
```

```python
new_df.head()
```

```python
y = new_df['Attrition']
X = new_df.drop('Attrition', axis = 1)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
scaled_df = scaler.fit_transform(X)
```

```python
scaled_df
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(scaled_df, y, test_size = 0.2, random_state = 42)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
print('Accuracy {} %'.format(accuracy_score(y_pred, y_test) * 100))
```

```python
cm = confusion_matrix(y_pred, y_test)
sns.heatmap(cm, annot=True, cmap='Blues')
plt.show()
```

```python
print(classification_report(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
forest = RandomForestClassifier()
```

```python
forest.fit(X_train, y_train)
```

```python
y_pred_forest = forest.predict(X_test)
```

```python
cm = confusion_matrix(y_pred_forest, y_test)
sns.heatmap(cm, annot=True, cmap='Blues')
plt.show()
```

```python
print('Accuracy {} %'.format(accuracy_score(y_pred_forest, y_test) * 100))
```

```python
print(classification_report(y_test, y_pred_forest))
```

```python
import tensorflow as tf

model_tf = tf.keras.models.Sequential()
```

```python
model_tf.add(tf.keras.layers.Dense(units = 500, activation = 'relu', input_shape = (X_train.shape[1],)))
model_tf.add(tf.keras.layers.Dense(units = 500, activation = 'relu'))
model_tf.add(tf.keras.layers.Dense(units = 500, activation = 'relu'))
model_tf.add(tf.keras.layers.Dense(units = 1, activation = 'sigmoid'))
```

```python
model_tf.summary()
```

```python
model_tf.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
epochs = model_tf.fit(X_train, y_train, epochs = 100, batch_size = 50)
```

```python
y_pred = model_tf.predict(X_test)
y_pred = (y_pred >0.5)
```

```python
plt.plot(epochs.history['loss'])
plt.title('Model Loss Progress During Training')
plt.xlabel('Epoch')
plt.ylabel('Training Loss')
plt.legend(['Training Loss'])
```

```python
plt.plot(epochs.history['accuracy'])
plt.title('Model Accuracy Progress During Training')
plt.xlabel('Epoch')
plt.ylabel('Training Accuracy')
plt.legend(['Training Accuracy'])
```

```python
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, cmap='Blues')
plt.show()
```

```python
print(classification_report(y_test, y_pred))
```
